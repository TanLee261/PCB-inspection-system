# 03 — Bounding-box size distribution
Chỉ đo các box có class ID hợp lệ, ảnh đọc được, chiều dương và nằm trong ảnh; các box lỗi xem notebook 02. Diện tích được đo theo pixel ảnh gốc, không resize. `relative_area = area_px / (image_width × image_height)`. Small: < 1024; medium: 1024–9216 (bao gồm hai đầu); large: > 9216.

```python
import json
from pathlib import Path
notebook_path = next((p / 'scripts' / '01_inventory.ipynb' for p in (Path.cwd(), *Path.cwd().parents) if (p / 'scripts' / '01_inventory.ipynb').is_file()), None)
if notebook_path is None:
    raise FileNotFoundError('01_inventory.ipynb not found: run inside the project.')
exec(compile(''.join(json.loads(notebook_path.read_text(encoding='utf-8'))['cells'][1]['source']), str(notebook_path), 'exec'))
```

```python
for name, result in all_audits().items():
    display(Markdown(f'## {name} — size distribution'))
    df = result['boxes'].copy()
    if df.empty:
        display(Markdown('No valid boxes to measure.'))
        continue
    df['size_bin'] = pd.cut(df['area_px'], bins=[-float('inf'), 1024, 9216, float('inf')], labels=['small', 'medium', 'large'], right=True)
    df.loc[df['area_px'] == 1024, 'size_bin'] = 'medium'
    display(Markdown('### Từng bounding box (đơn vị pixel, relative_area là tỷ lệ)'))
    display(df[['split', 'class', 'image', 'width_px', 'height_px', 'area_px', 'relative_area', 'size_bin']])
    display(Markdown('### Thống kê theo class'))
    display(df.groupby('class', sort=False)[['width_px', 'height_px', 'area_px', 'relative_area']].agg(['count', 'median', 'mean', 'min', 'max']).round(4))
    size_counts = df.groupby(['split', 'size_bin'], observed=False).size().unstack(fill_value=0).reindex(index=SPLITS, columns=['small', 'medium', 'large'], fill_value=0)
    size_counts.loc['TOTAL'] = size_counts.sum()
    display(Markdown('### Số lượng và tỷ lệ size bin theo split'))
    display(size_counts)
    display((size_counts.div(size_counts.sum(axis=1).replace(0, float('nan')), axis=0) * 100).round(2).add_suffix(' %'))
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    axes[0].hist(df['area_px'], bins=40)
    axes[0].set(title=f'{name}: box area histogram', xlabel='Area (px²)', ylabel='Boxes')
    df.boxplot(column='area_px', by='class', ax=axes[1], rot=35)
    axes[1].set(title=f'{name}: area by class', xlabel='Class', ylabel='Area (px²)')
    fig.suptitle('')
    plt.tight_layout()
    plt.show()
```
